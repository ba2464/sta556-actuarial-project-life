```python
import pandas as pd
from pathlib import Path

# Find the project root
current_dir = Path.cwd()

project_root = next(
    (
        path for path in [current_dir, *current_dir.parents]
        if (path / "_quarto.yml").exists()
    ),
    None
)

if project_root is None:
    raise FileNotFoundError("Project root not found.")

data_path = project_root / "data" / "raw" / "ssa_life_table_2023.html"

print("File exists:", data_path.exists())

if data_path.exists():
    tables = pd.read_html(data_path)

    print("Number of tables found:", len(tables))

    for i, table in enumerate(tables):
        print(f"\nTable {i}: {table.shape}")
        display(table.head())
```

```python
# Identify tables that appear to contain mortality data
for i, table in enumerate(tables):
    print(f"Table {i}: {table.shape}")

# Inspect the largest extracted table
mortality_raw = max(tables, key=len)

print("\nColumn structure:")
print(mortality_raw.columns)

print("\nFirst 10 observations:")
display(mortality_raw.head(10))

print("\nLast 10 observations:")
display(mortality_raw.tail(10))
```

```python

from pathlib import Path
import sys

# Find the repository root
project_root = next(
    (
        p for p in [Path.cwd(), *Path.cwd().parents]
        if (p / "data/raw/ssa_life_table_2023.html").is_file()
    ),
    None
)

if project_root is None:
    raise FileNotFoundError("Project root not found.")

# Allow Python to import files from src/
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.data import load_clean_mortality

# Load and clean the mortality data
data_path = (
    project_root
    / "data"
    / "raw"
    / "ssa_life_table_2023.html"
)

mortality, rows_removed = load_clean_mortality(data_path)

print("Cleaned dataset shape:", mortality.shape)
print("Rows removed:", rows_removed)

display(mortality.head(10))
display(mortality.tail(10))
```

```python

print("DATA QUALITY REPORT")
print("=" * 50)

print("\nNumber of observations:")
print(len(mortality))

print("\nNumber of variables:")
print(mortality.shape[1])

print("\nData types:")
print(mortality.dtypes)

print("\nMissing values:")
print(mortality.isna().sum())

print("\nDuplicate age-sex combinations:")
print(mortality.duplicated(["age", "sex"]).sum())

print("\nRecords by sex:")
print(mortality["sex"].value_counts())

print("\nAge range by sex:")
print(
    mortality.groupby("sex")["age"].agg(
        ["min", "max", "nunique"]
    )
)

print("\nDeath probability summary:")
display(
    mortality.groupby("sex")["qx"].describe()
)

print("\nSelected ages:")
display(
    mortality[
        mortality["age"].isin([0, 20, 40, 60, 80, 100])
    ]
)
```

```python
processed_dir = project_root / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

output_path = processed_dir / "ssa_mortality_clean.csv"

mortality.to_csv(output_path, index=False)

print("Cleaned data saved to:", output_path)
```
